```python
import pandas as pd
import matplotlib.pyplot as plt
from rolling_retention_code import compute_rolling_retention, plot_rolling_retention

df = pd.read_csv('helium_comments_df.csv')

# Compute rolling retention
results = compute_rolling_retention(
    df,
    base_unit='M',
    user_column="comment_author",
    date_column="comment_created_datetime",
    window_sizes=[1, 3, 6]  # 1 month, 3 months, 6 months back
)

# Plot all windows on one chart
fig, ax = plot_rolling_retention(results)
plt.savefig('helium_comments_rolling_retention_weekly.png', dpi=300, bbox_inches='tight')
plt.show()

# Check results
for window, data in results.items():
    mean_pct = data['stats']['mean_retention_rate'] * 100
    print(f"{window}-month: {mean_pct:.1f}% returning users")
```

```python
for window, data in results.items():
    print(data["retention_df"]["users_retained"].max())
```
